```python
# apache spark - distributed data processing engine
# in simple terms spark machine is used to process the large amount of data across multiple machines quickly
# spark is particularly useful for these large scale transformation
# for eg customer.csv has 100gb spark used to distribute the work load to servarl machinces like 30 40 40gb like that!

# why?
# pandas can process data in one machine but saprk can distribute across multiple machines

# what is pyspark
# python api for spark more like python way to use the spark

# spark's architecture
# 1.driver - main controller / runs code / tells executors what to do / create execution plan / collect the result
# 2. executors - do actually data processing
# 3. cluster - the environment containing them / the whole machines

# when we run a sprak operation it break into job > stages > tasks

# jobs - a job is created when u perform an action  / eg df.count() -> here count() is the action
# stages - group of operations or work
# tasks - smallest unit of the work partition

# lazy evaluation - spark doesnt execute transformation immediately it wait until the we perform actions
# eg
# df2 = df.filter(df["age"] > 25) # spark doesnt process the data yet it will remeber this operation
# df2.show() # this is an action now execute actually

# transformation - what to do
# a tranformation create a new dataframe but doesnt execute immediately
# df.filter(...)
# df.select(...)
# df.withColumn(...)
# df.groupBy(...)
# df.join(...)

# action - do it
#  an action tigger spark to execute the work
# df.show()
# df.count()
# df.collect()
# df.write.parquet(...)

# partition - small chunk of data
# spark splits the large data set into smaller pieces so that executors can process them in parallel
# each partition can be processed by task

# shuffle -  spark moves data btween partition/machines / its more like moving or reorganizing

```

```python
# PySpark groupBy() + agg()
# groupBy() - groups the rows by column
# agg() - calculate the agg value for the group

# remeber
# groupBy() - Groups rows
# count() - count rows in each group
# agg() - calculate multiple values (sum, min, max, avg, count)
# alias() - rename

# Pandas
# df.groupby("department")["salary"].sum()
# df.groupby("department")["salary"].agg(["sum", "mean", "max", "min"])

# PySpark
# emp.groupBy("department").agg(F.sum("salary"))

from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName("test").getOrCreate()

emp = spark.read.csv("/content/emp.csv", header=True, inferSchema=True)
emp.show(5)

# groupBy() - group the data
emp.groupBy("DEPARTMENT_ID").count().show() # This counts employees in each department.
emp.groupBy("DEPARTMENT_ID").max("SALARY").show()
emp.groupBy("DEPARTMENT_ID").min("SALARY").show()
emp.groupBy("DEPARTMENT_ID").avg("SALARY").show()
# in this example it count the values by the group we defined DEPARTMENT_ID

# agg() - calculate multiple values

emp.groupBy("DEPARTMENT_ID").agg(
    count("EMPLOYEE_ID").alias("EMPLOYEE_COUNT"),
    max("SALARY").alias("MAX_SALARY"),
    min("SALARY").alias("MIN_SALARY"),
    avg("SALARY").alias("AVG_SALARY"),
    sum("SALARY").alias("TOTAL_SALARY")
).show()

```

```python
# Pyspark common functions
# these functions help us to select calculate change and clean data

# first most important is
from pyspark.sql.functions import *
# or
from pyspark.sql import functions as F

# Both work, but F is generally preferred
# Why F? It makes it clear that the function comes from PySpark and avoids name conflicts.

# col() - reference a column
emp.select(F.col("FIRST_NAME"), F.col("SALARY")).show(1)
# lit() - create a fixed value
emp.select("FIRST_NAME", "SALARY").withColumn("COUNTRY", F.lit("INDIA")).show(5)
# when() + otherwise() - condition logiic like if else!
emp.select("FIRST_NAME", "SALARY").withColumn(
    "category",
    F.when(F.col("SALARY") > 10000, "HIGH")
    .when(F.col("SALARY") > 5000, "MEDIUM")
    .otherwise("LOW")
).show(5)
# cast() - change the data type!
emp.select("FIRST_NAME", "SALARY", F.col("SALARY").cast("double")).show(2)
```

```python
# Pyspark windows functions
# windows functions perform calculations across related rows without combining all rows into one

# row_number() → unique numbers, even for ties.
# rank() → ties share a rank; the next rank has a gap.
# dense_rank() → ties share a rank; no gaps.
# lag() - Previous row
# lead() - Next row

# to remember
# .partitionBy() - Define groups
# .orderBy() - Define row order
# row_number() - Unique row number
# rank() - Ranking with gaps
# dense_rank() - Ranking without gaps
# lag() - Previous row
# lead() - Next row

from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql import functions as F
from pyspark.sql.window import Window

spark = SparkSession.builder.appName("test").getOrCreate()

emp = spark.read.csv("/content/emp.csv", header=True, inferSchema=True)


# setup()
window_spec = Window.partitionBy("DEPARTMENT_ID").orderBy(F.col("SALARY").desc())
# emp.show() # to the see the output for the setup!
# partitionBy("") - creates a seprate groups
# orderBy() - sort


# row_number() - assigns a unique seqential number to each row within its window
emp.withColumn(
    "row_number",
    F.row_number().over(window_spec)
).show(5)

# rank()
emp.withColumn(
    "rank", F.rank().over(window_spec)
).show(5)

# dense_rank()
emp.withColumn(
    "dense_rank", F.dense_rank().over(window_spec)
).show(5)

# lag()
emp.withColumn(
    "previous_salary", F.lag("SALARY").over(window_spec)
).show(5)

# lead()
emp.withColumn(
    "next_salary", F.lead("SALARY").over(window_spec)
).show(5)

#
```

```python
# repartition() & coalesce()
# both are used to change the no of partions in spark

# repartition() - used to increase or decrease partitions
# use when we need to redistribute the data / causes shuffle -> data moves btw partitions -> can be expensive
emp = emp.repartition(2)

# coalesce() - mainly used to decrease partitions
df = df.coalesce(2)

# easy rule to remember
# Need to increase partitions?
# → repartition()

# Need to decrease partitions?
# → coalesce()

# Why is coalesce() generally faster than repartition() when reducing partitions?

# Answer: Because repartition() performs a shuffle, while coalesce() can reduce partitions without a full shuffle.
```

```python
from os.path import join
# Broadcast Join

# Used when joining a large dataset with a small dataset.

# Large + Large
→ Normal join
→ Shuffle may happen
→ Data is redistributed between partitions
→ Can be expensive

# Large + Small
→ Broadcast join
→ Small dataset is copied to executors
→ Avoids/reduces the need for a large shuffle
→ Usually faster

# example

result = emp.join(
    F.broadcast(dept),
    emp["DEPARTMENT_ID"] == dept["DEPARTMENT_ID"],
    "inner"
)
result.show()

# easy rule to remember
# Large + Large → normal join
# Large + Small → broadcast join
```
